# 🔧 Asistente Experto en Motores de Combustión
## Inteligencia Artificial aplicada al conocimiento técnico

---

Buenos días / Buenas tardes.

Hoy les voy a presentar un asistente de inteligencia artificial que convierte cientos de páginas de documentación técnica en respuestas instantáneas y precisas.

En lugar de que un experto pase horas buscando información en manuales, el sistema la encuentra en segundos — y recuerda el contexto de la conversación, igual que lo haría una persona.

**La presentación tiene 4 partes:**

- **El problema** que resuelve este asistente
- **La solución** que hemos construido
- **Demo en vivo** del asistente funcionando
- **Por qué este enfoque es diferente** y cómo escala

---

## 🎯 El Problema

---

Imaginen este escenario:

Un técnico necesita saber cómo funciona la fase de compresión de un motor de dos tiempos. Tiene cuatro manuales técnicos, más de 150 páginas, y no sabe en cuál está la respuesta.

**El resultado:** pierde tiempo, comete errores, o llama a un experto que tampoco lo sabe de memoria.

Este problema existe en cualquier industria donde el conocimiento está atrapado en documentos:

- 📄 Manuales técnicos
- 📋 Normativas y regulaciones
- 🏥 Protocolos médicos
- ⚖️ Documentación legal
- 🏗️ Ingeniería industrial

**La solución no es contratar más expertos. Es hacer que el conocimiento sea accesible al instante.**

---

## 💡 La Solución: El Asistente Experto

---

Lo que hemos construido es un **asistente de IA especializado** que:

✅ **Lee y comprende** cualquier documentación técnica

✅ **Responde preguntas** en lenguaje natural, basándose exclusivamente en esos documentos

✅ **Recuerda la conversación** — si preguntas sobre un tema y luego dices "¿y qué pasa después?", entiende a qué te refieres

✅ **No inventa** — si la respuesta no está en los documentos, lo dice claramente

✅ **Es resiliente** — si un motor de IA falla, automáticamente cambia a otro sin que el usuario lo note

En este caso concreto, el asistente es experto en **motores de combustión interna** — 4 manuales técnicos, 158 páginas, más de 600 fragmentos de conocimiento indexados y disponibles en milisegundos.

## ⚙️ Cómo Funciona — Sin tecnicismos

---

> **📺 [VOLVER AL NOTEBOOK]**

El sistema funciona en tres pasos:

**1. Indexación** — Cargamos los documentos y los convertimos en conocimiento estructurado. Es como crear un índice inteligente de todo el contenido. Se hace una sola vez.

**2. Búsqueda** — Cuando alguien hace una pregunta, el sistema encuentra automáticamente los fragmentos más relevantes. No busca por palabras clave como Google — entiende el significado.

**3. Respuesta** — Un modelo de lenguaje avanzado combina esos fragmentos con la pregunta y genera una respuesta coherente, clara y fundamentada.

```
Pregunta → Búsqueda inteligente → Contexto relevante → Respuesta precisa
```

Y si el modelo principal falla por sobrecarga, el sistema cambia automáticamente a un modelo de respaldo — sin interrupciones para el usuario.

## 🏆 Por Qué Este Enfoque

---

Hay tres decisiones de diseño clave que hacen este sistema robusto:

**Responde solo con lo que sabe** — Las respuestas están siempre ancladas a los documentos. Sin alucinaciones, sin información inventada.

**Memoria de conversación** — No trata cada pregunta de forma aislada. Entiende el contexto de la conversación completa.

**Resiliente por diseño** — Si el modelo de IA principal no está disponible, cambia automáticamente a alternativas sin que el usuario lo note.
